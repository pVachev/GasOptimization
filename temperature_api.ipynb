```python
# Import Meteostat library and dependencies
from datetime import date
import meteostat as ms

# Set time period
start = date(2025, 11, 1)
end = date(2025, 12, 31)

# Get daily data
ts = ms.daily(ms.Station(id='06590'), start, end)
df = ts.fetch()

# Print DataFrame
print(df['temp'])
```
